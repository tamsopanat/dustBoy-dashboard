```python
import pandas as pd
import numpy as np
```

```python
station_df = pd.read_csv('./station_df.csv')
station_df
```

```python
start_date = pd.to_datetime('2026-04-30')
dates = pd.date_range(start=start_date, periods=7)

all_days = []

for current_day in dates:
    temp_df = station_df.copy()
    temp_df['pm25'] = np.random.randint(20, 100, size=len(station_df))
    temp_df['date'] = current_day.date()
    all_days.append(temp_df)

# 3. Concat them all together at once
pred_station_df = pd.concat(all_days, ignore_index=True)
pred_station_df
```

```python
station_df_amphoe = pred_station_df[['amphoe', 'province', 'dustboy_lat', 'dustboy_lon', 'pm25', 'date']].groupby(by = ['date', 'province', 'amphoe'], as_index=False).mean()
station_df_amphoe = station_df_amphoe[station_df_amphoe['province'] != "ชียงใหม่"]
station_df_amphoe = station_df_amphoe[station_df_amphoe['amphoe'] != "มือง"]
station_df_amphoe.to_csv('./pm_mock.csv', index = False)
station_df_amphoe
```

```python
np.random.seed(42)
patient_data = []
unique_loc = station_df_amphoe[['province', 'amphoe']].drop_duplicates()

for _, row in unique_loc.iterrows():
    num_patients = np.random.randint(3, 8)  # Random 3-7 patients per station
    for i in range(num_patients):
        patient_data.append({
            'amphoe': row['amphoe'],
            'province': row['province'],
            'cid': f"1-{np.random.randint(1000, 9999)}-{np.random.randint(10000, 99999)}",
            'age': np.random.randint(5, 85),
            'sex': np.random.choice(['M', 'F']),
        })

# Create the final dataframe
df_patients = pd.DataFrame(patient_data)
df_patients.to_csv('./patient_mock.csv', index = False)
df_patients
```
